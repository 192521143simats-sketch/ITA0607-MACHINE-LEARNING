```python
# FIND-S Algorithm

data = [
    ['Sunny', 'Warm', 'Normal', 'Strong', 'Warm', 'Same', 'Yes'],
    ['Sunny', 'Warm', 'High', 'Strong', 'Warm', 'Same', 'Yes'],
    ['Rainy', 'Cold', 'High', 'Strong', 'Warm', 'Change', 'No'],
    ['Sunny', 'Warm', 'High', 'Strong', 'Cool', 'Change', 'Yes']
]
h = ['0', '0', '0', '0', '0', '0']

for row in data:
    if row[-1] == 'Yes':
        for i in range(len(h)):
            if h[i] == '0':
                h[i] = row[i]
            elif h[i] != row[i]:
                h[i] = '?'

print("Final hypothesis:")
print(h)
```

```python
#Candidate-Elimination

import pandas as pd

data = pd.DataFrame([
    ['Sunny', 'Warm', 'Normal', 'Strong', 'Warm', 'Same', 'Yes'],
    ['Sunny', 'Warm', 'High', 'Strong', 'Warm', 'Same', 'Yes'],
    ['Rainy', 'Cold', 'High', 'Strong', 'Warm', 'Change', 'No'],
    ['Sunny', 'Warm', 'High', 'Strong', 'Cool', 'Change', 'Yes']
])

S = ['0'] * 6
G = [['?', '?', '?', '?', '?', '?']]

for row in data.values:

    x = row[:-1]
    target = row[-1]

    if target == 'Yes':
        for i in range(6):
            if S[i] == '0':
                S[i] = x[i]
            elif S[i] != x[i]:
                S[i] = '?'

    else:
        new_G = []

        for g in G:
            for i in range(6):
                if g[i] == '?':
                    if S[i] != x[i]:
                        new_g = g.copy()
                        new_g[i] = S[i]
                        new_G.append(new_g)

        G = new_G

print("Specific Boundary S:")
print(S)

print("\nGeneral Boundary G:")
for g in G:
    print(g)
```

```python
from sklearn.tree import DecisionTreeClassifier
from sklearn.preprocessing import LabelEncoder

# Dataset
data = [
    ['Sunny', 'Hot', 'High', 'Weak', 'No'],
    ['Sunny', 'Hot', 'High', 'Strong', 'No'],
    ['Overcast', 'Hot', 'High', 'Weak', 'Yes'],
    ['Rain', 'Mild', 'High', 'Weak', 'Yes'],
    ['Rain', 'Cool', 'Normal', 'Weak', 'Yes'],
    ['Rain', 'Cool', 'Normal', 'Strong', 'No'],
    ['Overcast', 'Cool', 'Normal', 'Strong', 'Yes'],
    ['Sunny', 'Mild', 'High', 'Weak', 'No'],
    ['Sunny', 'Cool', 'Normal', 'Weak', 'Yes'],
    ['Rain', 'Mild', 'Normal', 'Weak', 'Yes']
]

X = [row[:-1] for row in data]
y = [row[-1] for row in data]

# Convert text into numbers
encoder = LabelEncoder()

X_encoded = []
for column in zip(*X):
    X_encoded.append(encoder.fit_transform(column))

X_encoded = list(zip(*X_encoded))

# ID3 uses entropy
model = DecisionTreeClassifier(criterion='entropy')
model.fit(X_encoded, y)

# New sample
new_sample = [['Sunny', 'Cool', 'High', 'Strong']]

encoded_sample = []

for i in range(4):
    values = [row[i] for row in X]
    e = LabelEncoder()
    e.fit(values)
    encoded_sample.append(e.transform([new_sample[0][i]])[0])

prediction = model.predict([encoded_sample])

print("Prediction:", prediction[0])
```

```python
import numpy as np

# XOR input
X = np.array([
    [0, 0],
    [0, 1],
    [1, 0],
    [1, 1]
])

y = np.array([
    [0],
    [1],
    [1],
    [0]
])

# Sigmoid function
def sigmoid(x):
    return 1 / (1 + np.exp(-x))

def derivative(x):
    return x * (1 - x)

# Random weights
np.random.seed(1)

w1 = np.random.rand(2, 2)
w2 = np.random.rand(2, 1)

# Training
for epoch in range(10000):

    # Forward propagation
    hidden = sigmoid(np.dot(X, w1))
    output = sigmoid(np.dot(hidden, w2))

    # Error
    error = y - output

    # Backpropagation
    d_output = error * derivative(output)
    d_hidden = np.dot(d_output, w2.T) * derivative(hidden)

    # Update weights
    w2 += np.dot(hidden.T, d_output)
    w1 += np.dot(X.T, d_hidden)

print("Predicted output:")
print(np.round(output))
```

```python
from sklearn.neighbors import KNeighborsClassifier

# Training data
X = [
    [1, 2],
    [2, 3],
    [3, 3],
    [6, 7],
    [7, 8],
    [8, 7]
]

y = ['A', 'A', 'A', 'B', 'B', 'B']

# Create KNN model
model = KNeighborsClassifier(n_neighbors=3)

model.fit(X, y)

# Test sample
sample = [[5, 6]]

prediction = model.predict(sample)

print("Predicted class:", prediction[0])
```
