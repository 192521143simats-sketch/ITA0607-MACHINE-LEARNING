```python
import numpy as np
from sklearn.mixture import GaussianMixture

# Data
X = np.array([
    [1], [2], [3],
    [10], [11], [12]
])

# EM algorithm
model = GaussianMixture(n_components=2, random_state=42)

model.fit(X)

# Find clusters
clusters = model.predict(X)

print("Cluster for each data point:")
print(clusters)

print("\nMeans:")
print(model.means_)
```
