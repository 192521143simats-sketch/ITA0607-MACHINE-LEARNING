```python
from sklearn.tree import DecisionTreeClassifier

# Features:
# Income, Age, Existing Loan
X = [
    [25000, 22, 1],
    [30000, 25, 0],
    [50000, 30, 0],
    [70000, 35, 0],
    [20000, 21, 1],
    [80000, 40, 0]
]

# 0 = Poor credit
# 1 = Good credit
y = [0, 0, 1, 1, 0, 1]

model = DecisionTreeClassifier()

model.fit(X, y)

# New person
new_person = [[60000, 32, 0]]

prediction = model.predict(new_person)

if prediction[0] == 1:
    print("Good Credit Score")
else:
    print("Poor Credit Score")
```
