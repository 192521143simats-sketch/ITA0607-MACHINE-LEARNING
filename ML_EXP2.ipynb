```python
import pandas as pd

data = pd.DataFrame([
    ['Sunny', 'Warm', 'Normal', 'Strong', 'Warm', 'Same', 'Yes'],
    ['Sunny', 'Warm', 'High', 'Strong', 'Warm', 'Same', 'Yes'],
    ['Rainy', 'Cold', 'High', 'Strong', 'Warm', 'Change', 'No'],
    ['Sunny', 'Warm', 'High', 'Strong', 'Cool', 'Change', 'Yes']
])

S = ['0'] * 6
G = [['?', '?', '?', '?', '?', '?']]

for row in data.values:

    x = row[:-1]
    target = row[-1]

    if target == 'Yes':
        for i in range(6):
            if S[i] == '0':
                S[i] = x[i]
            elif S[i] != x[i]:
                S[i] = '?'

    else:
        new_G = []

        for g in G:
            for i in range(6):
                if g[i] == '?':
                    if S[i] != x[i]:
                        new_g = g.copy()
                        new_g[i] = S[i]
                        new_G.append(new_g)

        G = new_G

print("Specific Boundary S:")
print(S)

print("\nGeneral Boundary G:")
for g in G:
    print(g)
```
