```python
from sklearn.neighbors import KNeighborsClassifier

# Training data
X = [
    [1, 2],
    [2, 3],
    [3, 3],
    [6, 7],
    [7, 8],
    [8, 7]
]

y = ['A', 'A', 'A', 'B', 'B', 'B']

# Create KNN model
model = KNeighborsClassifier(n_neighbors=3)

model.fit(X, y)

# Test sample
sample = [[5, 6]]

prediction = model.predict(sample)

print("Predicted class:", prediction[0])
```
